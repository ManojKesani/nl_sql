```python
import os
import json
from dotenv import load_dotenv
from pprint import pprint

load_dotenv()

from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain.messages import HumanMessage,SystemMessage,AIMessage

# print(os.environ['GROQ_API_KEY'])

llm = ChatGroq(
    model = 'qwen/qwen3.8-27b'
)

agent = create_agent(model = llm)

def load_msg_from_json(json_file):
    with open(json_file,"r") as f:
        data = json.load(f)
    return data['content']


msg = load_msg_from_json('input.json')
system_msg = SystemMessage("You are a helpful coding assistant.")
human_msg = HumanMessage(content=msg)
messages = [
    system_msg,
    human_msg    
]
# print(message)

result = agent.invoke({"messages": messages})
pprint(result)

# Convert each message object using .model_dump()
output_data = {
    "messages": [m.model_dump() for m in result.get("messages", [])]
}

# Save result to output.json
with open("output.json", "w") as f:
    json.dump(output_data, f, indent=4, default=str)

print("Result successfully saved to output.json")
```
